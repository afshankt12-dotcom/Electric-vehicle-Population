### import data

### Check missing values

```python

print(df.isnull().sum())
```

# 1. Remove duplicates

```python
df = df.drop_duplicates()
```

# 2. Handle missing values

```python
df = df.dropna()
```

# 3. Fix column names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

### 4. Remove unwanted columns

```python
df = df.drop(columns=['unwanted_column1', 'unwanted_column2'], errors='ignore')
```

```python
print(df.columns)
```

```python
df.to_csv('Electric_Vehicle_Population_Data.csv', index=False)
```
